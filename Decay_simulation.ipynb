```python
import numpy as np
import matplotlib.pyplot as plt

# Half-lives (in years)
#half_life_U238 = 4.468e9           # U-238
half_life_Co60 = 5.27              # Co-60

# Decay constants
#lambda_U238 = np.log(2) / half_life_U238
lambda_Co60 = np.log(2) / half_life_Co60

# Time range for simulation
t = np.linspace(0, 10 * half_life_Co60, 500)  # years

# Initial number of nuclei
N0 = 1e6

# Decay equations
#N_U238 = N0 * np.exp(-lambda_U238 * t)
N_Co60 = N0 * np.exp(-lambda_Co60 * t)

# Plot
plt.figure(figsize=(10,6))
#plt.plot(t, N_U238, label="U-238")
plt.plot(t, N_Co60, label="Co-60")
plt.xlabel("Time (years)")
plt.ylabel("Number of nuclei")
plt.title("Radioactive Decay of U-238 and Co-60")
plt.legend()
plt.grid(True)
plt.show()
```
